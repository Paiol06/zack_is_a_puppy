```python
import pandas as pd
```

```python
dataset = pd.read_csv('Gaming and Mental Health.csv', index_col = 0)
dataset.head(5)
```

**set index**

**duplicated records**

```python
dataset[dataset.duplicated(keep = False)].shape
```

**missing values**

```python
dataset.isna().sum()
```

```python
dataset.drop('work_productivity_score', axis = 1, inplace = True)
```

```python
numeric_columns = dataset.select_dtypes(include=['float64', 'int64']).columns
```

```python
dataset[numeric_columns] = dataset[numeric_columns].fillna(dataset[numeric_columns].mean())
```

```python
dataset.isna().sum()
```

```python
#do a boxplot for every column to check for outliers
import matplotlib.pyplot as plt
for column in numeric_columns:
    plt.figure()
    dataset.boxplot(column=column)
    plt.title(column)
    plt.show()
    
```
